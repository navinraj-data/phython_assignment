```python
print('Number guessing game')
import random

secret_number = random.randint(1, 10)
attempts = 3

while attempts > 0:
    guess = int(input("Enter your number to guess: "))

    # Out of range
    if guess < 1 or guess > 10:
        print("Your guess is out of range (1 to 10)")
        continue

    # Correct guess
    if guess == secret_number:
        print("Congratulations! You are correct 🎉")
        break

    # Wrong guess
    attempts -= 1

    if guess > secret_number:
        print(f"Too high! Attempts left: {attempts}")
    else:
        print(f"Too low! Attempts left: {attempts}")

# Runs if loop ends without break
else:
    print("Better luck next time! The secret number was:", secret_number)
```

```python
# Take input from user
number = int(input("Enter the number for which you want the multiplication table: "))

# Generate table using for loop
for i in range(1, 11):
    result = number * i
    print(f"{number} x {i} = {result}")
```

```python
# Function to calculate BMI
def calculate_bmi(weight, height):
    return weight / (height ** 2)

# User input
weight = float(input("Enter your weight in kg: "))
height = float(input("Enter your height in meters: "))

# Calculate BMI
bmi = calculate_bmi(weight, height)

# Display result
print(f"Your BMI is: {bmi:.2f}")
```
