```python
print('c. Create two sets: set1 with values: 1, 3, 5, 7, 9 set2 with values: 2, 3, 5, 8, 10')
set1 = {1, 3, 5, 7, 9}
set2 = {2, 3, 5, 8, 10}
print(set1)
print(set2)
```

```python
print('d. Compute and print the union and intersection of set1 and set2.')
union_set = set1.union(set2)
intersection_set = set1.intersection(set2)
print('Union of set1 and set2:', union_set)
print('Intersection of set1 and set2:', intersection_set)
```
